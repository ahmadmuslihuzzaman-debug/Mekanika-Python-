Pesawat Ulang Alik dengna Hambatan udara

```python
import numpy as np
import matplotlib.pyplot as plt


# Parameter :
f = 35e6 # gaya pada pesawat (N)
m = 2e6 # massa pesawat ulang alik (Kg)
g    = 9.81 # gravitasi bumi (m/s2)
v0 = 0.0 # Kecepatan pesawat (m/s)
y0 = 0.0 # Gaya berat kebawah
t_akhir = 20.0
t0 = 0.0
dt = 0.01 # langkah waktu (s)
n_step = 20000

# persamaan analitik
w = m*g
a = (f - w)/m
v = v0 + (a*t_akhir)
y = y0 + (v0*t_akhir) + (1/2 *a*t_akhir**2)

def Euler_pos(v0, y0, t0):
    v = v0 + (a*dt)
    y = y0 + (v0*dt)
    t = t0 + dt
    return v, y, t
v_graf = np.zeros(n_step)
y_graf = np.zeros(n_step)
t_graf = np.zeros(n_step)

for i in range(1, n_step):
  v, y, t = Euler_pos(v0, y0, t0)
  v0 = v
  y0 = y
  t0 = t

  # Menyimpan nilai v, y, t
  v_graf[i] = v
  y_graf[i] = y
  t_graf[i] = t
print(v, y, t)
```

```python
plt.figure(1)
plt.plot(t_graf[:i+1], v_graf[:i+1],label = "Kecepatan pesawat ulang alik")
plt.legend()
plt.xlabel("waktu (s)")
plt.ylabel("kecepatan (m/s)")
plt.title("Grafik kecepatan pesawat ulang alik")
plt.grid(True)
plt.show()
```
