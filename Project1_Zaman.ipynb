**Proyek1_ Menghitung Kecepatan Benda Akibat Gesekan**

```python
# model :
  # a = -mu*g {jika v > 0}
  # a = 0     {jika v = 0}
  # a = +mu*g {jika v < 0}
# parameter :
  # v0 = 5.0  kecepatan awal (m/s)
  # g = 9.81 gaya gravitasi (m/s2)
  # mu = 0.1  gaya gesek

import numpy as np
import matplotlib.pyplot as plt

v0 = 5.0 # kecepatan awal (m/s)
g = 9.81 # gaya gravitasi (m/s2)
mu = 0.1 # koefesien gesekan
dt = 0.01 # langkah waktu (s)

# waktu simulasi :
t_stop = v0 / (mu*g)

print(f"waktu berhenti :{t_stop:.3f} detik")

s0 = 0.0 # jarak awal (m)
dt = 0.01 # langkah waktu (s)
t0 = 0.0 # Waktu awal (s)
a = mu*g  # Percepatan (m/s)
n_step = 800
def Euler_pos(v0, a, dt, s0, t0):
  v = v0 - a*dt
  s = s0 + v0*dt
  t = t0 + dt
  return s, v, t
v_graf = np.zeros(n_step)
s_graf = np.zeros(n_step)
t_graf = np.zeros(n_step)
v_graf[0]= v0

for i in range(1, n_step):
  s, v, t = Euler_pos(v0, a, dt, s0, t0)
  s0 = s
  v0 = v
  t0 = t

  # Menyimpan nilai s, v, t
  v_graf[i] = v
  s_graf[i] = s
  t_graf[i] = t
  if v<=0:
    break

print(s, v, t, i)
```

```python
plt.figure(1)
plt.plot(t_graf[:i+1], v_graf[:i+1])
plt.show()
```
